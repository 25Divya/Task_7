```python
import pandas as pd
import sqlite3
# Load CSV
df = pd.read_csv("/content/drive/MyDrive/1000 Sales Records.csv")
# Create connection and table
conn = sqlite3.connect("sales_data.db")
df.to_sql("sales", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT
    `Item Type` AS product,
    SUM(`Units Sold`) AS total_qty,
    SUM(`Units Sold` * `Unit Price`) AS revenue
FROM sales
GROUP BY `Item Type`
"""

df_summary = pd.read_sql_query(query, conn)
```

```python
import matplotlib.pyplot as plt

# Print summary
print(df_summary)

# Plot bar chart
df_summary.plot(kind='bar', x='product', y='revenue', legend=False, color='skyblue')
plt.title("Total Revenue by Product")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("sales_chart.png")
plt.show()
```
